## Helper functions

```python
def add_paths(comparisons, window_dir, bigwig_dir, bam_dir):
    comparisons = comparisons.copy()

    for side in ["A", "B"]:
        exp = comparisons[f"experiment_{side}"]

        comparisons[f"window_{side}"] = exp.apply(
            lambda x: f"{window_dir}/{x}.reproducible_enriched_windows.tsv.gz")

        comparisons[f"plus_1_{side}"] = exp.apply(
            lambda x: f"{bigwig_dir}/plus/{x}_IP_1.unscaled.plus.bw")
        comparisons[f"plus_2_{side}"] = exp.apply(
            lambda x: f"{bigwig_dir}/plus/{x}_IP_2.unscaled.plus.bw")
        comparisons[f"minus_1_{side}"] = exp.apply(
            lambda x: f"{bigwig_dir}/minus/{x}_IP_1.unscaled.minus.bw")
        comparisons[f"minus_2_{side}"] = exp.apply(
            lambda x: f"{bigwig_dir}/minus/{x}_IP_2.unscaled.minus.bw")

        comparisons[f"plus_IN_1_{side}"] = exp.apply(
            lambda x: f"{bigwig_dir}/plus/{x}_IN_1.unscaled.plus.bw")
        comparisons[f"minus_IN_1_{side}"] = exp.apply(
            lambda x: f"{bigwig_dir}/minus/{x}_IN_1.unscaled.minus.bw")

        comparisons[f"bam_1_{side}"] = exp.apply(
            lambda x: f"{bam_dir}/{x}_IP_1.genome.Aligned.sort.dedup.bam")
        comparisons[f"bam_2_{side}"] = exp.apply(
            lambda x: f"{bam_dir}/{x}_IP_2.genome.Aligned.sort.dedup.bam")
        comparisons[f"bam_IN_1_{side}"] = exp.apply(
            lambda x: f"{bam_dir}/{x}_IN_1.genome.Aligned.sort.dedup.bam")

    return(comparisons)
```

# Running

## Actual run

```python
comparisons = add_paths(comparisons, window_dir=WINDOW_DIR, bigwig_dir=BIGWIG_DIR, bam_dir = BAM_DIR)
```

## Checking

```python
path_columns = [
    column for column in comparisons.columns
    if column.startswith("window_") or "bam" in column or "plus" in column or "minus" in column
]

missing = []

for column in path_columns:
    for path in comparisons[column]:
        if not os.path.exists(path):
            missing.append((column, path))

print(f"{len(missing)} missing files")

for column, path in missing:
    print(f"{column}: {path}")
```
