## Helper functions

```python
def add_bigwig_paths(comparisons):
    comparisons = comparisons.copy()

    for side in ["A", "B"]:
        exp = comparisons[f"experiment_{side}"]

        comparisons[f"plus_1_{side}"] = exp.apply(
            lambda x: f"{BIGWIG_DIR}/plus/{x}_IP_1.scaled.plus.bw")
        comparisons[f"plus_2_{side}"] = exp.apply(
            lambda x: f"{BIGWIG_DIR}/plus/{x}_IP_2.scaled.plus.bw")
        comparisons[f"minus_1_{side}"] = exp.apply(
            lambda x: f"{BIGWIG_DIR}/minus/{x}_IP_1.scaled.minus.bw")
        comparisons[f"minus_2_{side}"] = exp.apply(
            lambda x: f"{BIGWIG_DIR}/minus/{x}_IP_2.scaled.minus.bw")

    return(comparisons)
```

```python
WINDOW_DIR = "/tscc/projects/ps-yeolab5/rbp-portal/encode3/eclip/Skipper/2.0.0/v1/output/reproducible_enriched_windows"

def add_input_paths(comparisons):
    comparisons = comparisons.copy()

    for side in ["A", "B"]:
        exp = comparisons[f"experiment_{side}"]

        comparisons[f"window_{side}"] = exp.apply(
            lambda x: f"{WINDOW_DIR}/{x}.reproducible_enriched_windows.tsv.gz")

        comparisons[f"plus_1_{side}"] = exp.apply(
            lambda x: f"{BIGWIG_DIR}/plus/{x}_IP_1.scaled.plus.bw")
        comparisons[f"plus_2_{side}"] = exp.apply(
            lambda x: f"{BIGWIG_DIR}/plus/{x}_IP_2.scaled.plus.bw")
        comparisons[f"minus_1_{side}"] = exp.apply(
            lambda x: f"{BIGWIG_DIR}/minus/{x}_IP_1.scaled.minus.bw")
        comparisons[f"minus_2_{side}"] = exp.apply(
            lambda x: f"{BIGWIG_DIR}/minus/{x}_IP_2.scaled.minus.bw")

    return(comparisons)
```

# Running

## Actual run

```python
comparisons = add_bigwig_paths(comparisons)
comparisons = add_input_paths(comparisons)
```

## Checking

```python
path_columns = ["window_A", "window_B"] + [
    f"{strand}_{rep}_{side}"
    for side in ["A", "B"]
    for strand in ["plus", "minus"]
    for rep in [1, 2]
]

missing = []

for column in path_columns:
    for path in comparisons[column]:
        if not os.path.exists(path):
            missing.append(path)

print(f"{len(missing)} missing files")

for path in missing:
    print(path)
```
